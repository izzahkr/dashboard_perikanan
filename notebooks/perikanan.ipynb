```python
import os
import glob
import pandas as pd

folder_path = "DATA"

# ============================================================
# 2. MENGAMBIL SEMUA FILE EXCEL DAN CSV
# ============================================================
file_patterns = [
    "*.xlsx",
    "*.xls",
    "*.csv"
]

file_list = []

for pattern in file_patterns:
    file_list.extend(glob.glob(os.path.join(folder_path, pattern)))

print(f"Total file ditemukan: {len(file_list)}")
print("=" * 80)

# Dictionary untuk menyimpan dataframe
data_dict = {}

# ============================================================
# 3. MEMBACA SEMUA FILE
# ============================================================
for file_path in file_list:

    file_name = os.path.basename(file_path)
    file_ext = os.path.splitext(file_name)[1].lower()

    print(f"\n📁 FILE: {file_name}")
    print("-" * 80)

    try:

        # ====================================================
        # JIKA FILE CSV
        # ====================================================
        if file_ext == ".csv":

            # Coba membaca CSV dengan separator koma
            try:
                df = pd.read_csv(file_path)
            
            except:
                # Jika gagal, coba separator titik koma
                df = pd.read_csv(file_path, sep=";")

            data_dict[file_name] = df

            print("📄 [CSV]")
            print(f"Ukuran (Baris x Kolom): {df.shape}")
            print("Kolom:", list(df.columns))

            print("\n🔹 5 BARIS PERTAMA:")
            display(df.head(5))

            print("\n🔹 1 BARIS TERAKHIR:")
            display(df.tail(1))

        # ====================================================
        # JIKA FILE EXCEL
        # ====================================================
        else:

            excel_file = pd.ExcelFile(file_path)
            sheet_names = excel_file.sheet_names

            print(f"Sheet yang ada: {sheet_names}")

            for sheet in sheet_names:

                df = pd.read_excel(
                    file_path,
                    sheet_name=sheet
                )

                # Simpan dataframe
                data_dict[f"{file_name} - {sheet}"] = df

                print(f"\n📄 [Sheet: {sheet}]")
                print(f"Ukuran (Baris x Kolom): {df.shape}")
                print("Kolom:", list(df.columns))

                print("\n🔹 5 BARIS PERTAMA:")
                display(df.head(5))

                print("\n🔹 1 BARIS TERAKHIR:")
                display(df.tail(2))

                print("-" * 60)

    except Exception as e:

        print(f"❌ Gagal membaca file {file_name}")
        print(f"Error: {e}")

# ============================================================
# 4. SELESAI
# ============================================================
print("\n" + "=" * 80)
print("Selesai membaca seluruh file!")
print(f"Total dataframe yang berhasil disimpan: {len(data_dict)}")
```
